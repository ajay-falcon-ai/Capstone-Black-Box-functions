```python
import os

# List the contents of the parent directory to see how 'Black-Box Functions' is represented
parent_path = '/content/drive/MyDrive/Capstone'

if os.path.exists(parent_path):
    print(f"Contents of '{parent_path}':")
    for item in os.listdir(parent_path):
        item_full_path = os.path.join(parent_path, item)
        item_type = "(dir)" if os.path.isdir(item_full_path) else "(file)"
        print(f"- {item} {item_type}")
else:
    print(f"Error: Parent directory '{parent_path}' not found.")
```
