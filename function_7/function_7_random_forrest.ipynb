```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor

# -------------------------------
# Step 1: Utility functions
# -------------------------------

def train_random_forest(df, feature_cols, target_col):
    """Train a RandomForestRegressor on rows with non-NaN target values."""
    train_df = df.dropna(subset=[target_col])
    X_train = train_df[feature_cols]
    y_train = train_df[target_col]

    model = RandomForestRegressor(n_estimators=200, random_state=42)
    model.fit(X_train, y_train)
    return model


def predict_missing(df, model, feature_cols, target_col):
    """Predict missing target values and return updated DataFrame."""
    missing_df = df[df[target_col].isna()]
    if not missing_df.empty:
        predictions = model.predict(missing_df[feature_cols])
        df.loc[missing_df.index, target_col] = predictions
    return df


def plot_feature_importance(model, feature_cols, title="Feature Importance"):
    """Plot feature importances from the trained model."""
    importances = model.feature_importances_
    plt.figure(figsize=(8, 5))
    plt.barh(feature_cols, importances, color="skyblue")
    plt.xlabel("Importance")
    plt.title(title)
    plt.show()


def plot_actual_vs_predicted(df, model, feature_cols, target_col):
    """Scatter plot of actual vs predicted values for rows with known target."""
    known_df = df.dropna(subset=[target_col])
    X_known = known_df[feature_cols]
    y_known = known_df[target_col]
    y_pred = model.predict(X_known)

    plt.figure(figsize=(6, 6))
    plt.scatter(y_known, y_pred, alpha=0.7, color="darkorange")
    plt.plot([y_known.min(), y_known.max()],
             [y_known.min(), y_known.max()],
             "k--", lw=2)  # diagonal line
    plt.xlabel("Actual")
    plt.ylabel("Predicted")
    plt.title("Actual vs Predicted Scores")
    plt.show()


# -------------------------------
# Step 2: Example usage
# -------------------------------

if __name__ == "__main__":
    # Replace with your dataset loading
    # Example: df = pd.read_csv("your_data.csv")
    df = pd.DataFrame({
        # paste your dataset here
    })

    feature_cols = [col for col in df.columns if col.startswith("hype-para")]
    target_col = "score"

    # Train model
    model = train_random_forest(df, feature_cols, target_col)

    # Predict missing values
    df = predict_missing(df, model, feature_cols, target_col)
    print("Predicted missing scores:")
    print(df.loc[df.index[df[target_col].isna() == False], [target_col]])

    # Plot feature importance
    plot_feature_importance(model, feature_cols, title="Feature Importance for Score Prediction")

    # Plot actual vs predicted
    plot_actual_vs_predicted(df, model, feature_cols, target_col)
```
