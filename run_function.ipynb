```python
# Single-cell notebook content (paste into a new notebook cell)
%matplotlib inline

import sys
import os
from pathlib import Path
import matplotlib as mpl

# Resolve project root so run_function can be imported
try:
    project_root = os.path.abspath(os.path.join(os.path.dirname(__file__), ".."))
except NameError:
    project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

# Import helper that prepares config and returns runner info
from tools import notebook_launcher

print("Notebook dir:", Path.cwd())

# 1) Launch to get merged config and runner path (unchanged)
rc, tmp_config, runner_path = notebook_launcher.launch_from_notebook(notebook_dir=Path.cwd())
print("Launcher returned rc:", rc)
print("Merged config path:", tmp_config)
print("Runner used:", runner_path)

# 2) Print merged config contents
print("\n--- Merged config contents ---")
try:
    print(Path(tmp_config).read_text())
except Exception as e:
    print("Could not read merged config:", e)

# 3) Import the runner module and call its functions directly (Option 1)
#    This runs the pipeline inside the notebook kernel so plots created
#    by PlotUtils (plt.show()) will render inline.
from run_function import load_config, instantiate_data_handler, run_pipeline

cfg = load_config(tmp_config)

plot_cfg = cfg.get("plot", {})
figsize = tuple(plot_cfg.get("figsize", (12, 8)))
mpl.rcParams["figure.figsize"] = figsize

# Allow CLI-style overrides if notebook_launcher provided them in cfg
# (not required; kept for parity with original behavior)
if "out_dir" in cfg and cfg["out_dir"] is not None:
    print("Using out_dir from merged config:", cfg["out_dir"])

# Instantiate data handler in-kernel
dh = instantiate_data_handler(cfg["data_module"], cfg["data_class"])

# Run the pipeline in the same process (plots will display inline)
results = run_pipeline(cfg, dh)

# Optionally inspect results
print("\nPipeline finished. Results keys:", list(results.keys()))
```
