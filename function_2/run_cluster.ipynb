```python
import sys, os

project_root = os.path.abspath(os.path.join(os.getcwd(), ".."))
sys.path.append(project_root)

import pandas as pd

pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', 0)                  # Use unlimited width (auto-detect terminal size)
pd.set_option('display.expand_frame_repr', False)  # Prevent wrapping to multiple lines

from clustering.run_clustering import run_clustering
from clustering.cluster_analysis import ClusterReport
from clustering.handler_factory import get_handler

function_name = "function_2"

labels, model, df, df_scaled = run_clustering(function_name, method="kmeans", n_clusters=4)
handler = get_handler(function_name)
print(f"Clustering Report for function: {function_name}, method: kmean, n_clusters: 4")
report = ClusterReport(df, labels, handler.output_label, function_name)
report.generate()

labels, model, df, df_scaled = run_clustering(function_name, method="hierarchical", n_clusters=3)
handler = get_handler(function_name)
print(f"Clustering Report for function: {function_name}, method: hierarchical, n_clusters: 3")
report = ClusterReport(df, labels, handler.output_label, function_name)
report.generate()

labels, model, df, df_scaled = run_clustering(function_name, method="hierarchical", n_clusters=-1)
handler = get_handler(function_name)
print(f"Clustering Report for function: {function_name}, method: hierarchical, n_clusters: -1")
report = ClusterReport(df, labels, handler.output_label, function_name)
report.generate()
```
